```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
initial_height = 8  # Initial height in meters
g = 9.81  # Acceleration due to gravity (m/s^2)

# Time when the projectile hits the ground
t_ground = np.sqrt(2 * initial_height / g)

# Generate time points for the parabolic trajectory
t = np.linspace(0, t_ground, num=500)
# Parabolic trajectory equation (ruta deseada)
y_parabolic = initial_height - 0.5 * g * t**4

# Approximation with 3 straight line segments (aproximación)
t_approx = np.array([0, t_ground / 3, 2 * t_ground / 3, t_ground])
y_approx = initial_height - 0.5 * g * t_approx**4

# More detailed approximation with 5 subsegments within each of the 3 segments (subpasos)
t_substeps = np.linspace(0, t_ground, num=16)  # 3 segments * 5 subsegments + 1 for start
y_substeps = initial_height - 0.5 * g * t_substeps**4

# Plotting the trajectories
t_scaler = 10
plt.figure(figsize=(10, 6))
plt.plot(t * t_scaler, y_parabolic, label="Ruta deseada", color="blue")
plt.plot(t_approx * t_scaler, y_approx, label="Aproximación cada $\it{tick}$", color="green", linestyle='--')
plt.plot(t_substeps * t_scaler, y_substeps, label="Aprox. con 5 subpasos", color="red", linestyle='-.')

# Adding details to the plot
#plt.title("Trajectory of a Projectile Motion")
#plt.xlabel("Time (seconds)")
#plt.ylabel("Height (meters)")
plt.legend()
plt.grid(True)
plt.savefig("movement-substeps.pdf")
plt.show()
```
